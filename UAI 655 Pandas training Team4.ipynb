## Bonus: Sample Data Generation Code

```python
import pandas as pd
import numpy as np

# For Set 1 - Protein expression
np.random.seed(42)
proteins = [f'Protein_{i}' for i in range(1, 21)]
compartments = np.random.choice(['Nucleus', 'Cytosol', 'Mitochondria', 'Membrane'], 20)
df_protein = pd.DataFrame({
    'Protein': proteins,
    'Control': np.random.lognormal(2, 0.5, 20),
    'Treatment_A': np.random.lognormal(2.5, 0.6, 20),
    'Treatment_B': np.random.lognormal(1.8, 0.4, 20),
    'MW_kDa': np.random.uniform(10, 150, 20),
    'Cellular_Compartment': compartments
})
```

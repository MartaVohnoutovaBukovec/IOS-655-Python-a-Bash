# 1. Introduction – Why Command Line?

## What is CLI?
- **CLI** = Command Line Interface

- Text-based interaction with the operating system

- Faster, more powerful, and scriptable than GUI

## Why learn it?
- You control the system directly

- Essential for servers, DevOps, cloud, embedded systems

- Automate repetitive tasks

- "Graphical interface is for users, command line is for programmers"

# 2. First Steps – Basic Commands
## Your first commands
```bash
whoami              # who you are
pwd                 # print working directory
date                # current date and time
cal                 # calendar
echo "Hello World"  # print something
clear               # clean the screen (or Ctrl+L)
Getting help
bash
man ls              # manual for ls command
ls --help           # short help
whatis ls           # one-line description
```

# 3. Navigation – Moving Around
## Listing files
```bash
ls                  # list files
ls -l               # long format (details)
ls -a               # show hidden files (dot files)
ls -la              # combine options
ls -lh              # human-readable sizes
```
## Changing directories
```bash
cd /home            # go to absolute path
cd Documents        # go to relative path (from where you are)
cd ..               # go one level up
cd ~                # go to home directory
cd -                # go to previous directory
```
## Understanding paths
**Absolute path** – starts with / (e.g., /home/user/Documents)

**Relative path** – from current position (e.g., Documents/notes.txt)

. = current directory

.. = parent directory

~ = home directory


# 4. Working with Files and Directories
## Creating
```bash
touch file.txt              # create empty file
mkdir myfolder              # create directory
mkdir -p parent/child       # create parent and child
```
## Copying, moving, renaming, deleting
```bash
cp file.txt copy.txt        # copy file
cp -r folder newfolder      # copy directory recursively
mv file.txt newname.txt     # rename/move
rm file.txt                 # delete file (careful!)
rm -r folder                # delete directory recursively
rm -rf folder               # force delete (DANGEROUS!)
```
## Viewing files
```bash
cat file.txt                # print whole file
less file.txt               # page through file (q to quit)
head file.txt               # first 10 lines
head -n 20 file.txt         # first 20 lines
tail file.txt               # last 10 lines
tail -f log.txt             # follow file (live updates)
```

# 5. File Content – Searching and Processing
## Searching with ***grep***
```bash
grep "error" log.txt        # find lines containing "error"
grep -i "error" log.txt     # case insensitive
grep -r "TODO" .            # recursive search in all files
grep -v "debug" log.txt     # lines NOT containing "debug"
grep "^start" file.txt      # lines starting with "start"
grep "end$" file.txt        # lines ending with "end"
```
## Counting, sorting, unique
```bash
wc -l file.txt              # count lines
wc -w file.txt              # count words
wc -c file.txt              # count characters
sort names.txt              # sort alphabetically
sort -n numbers.txt         # sort numerically
uniq                        # remove duplicates (requires sort first)
sort file.txt | uniq        # unique lines
sort file.txt | uniq -c     # count occurrences
```

# 6. Redirection and Pipes
## Output redirection ***(>, >>)***
```bash
ls > files.txt              # save output to file (overwrite)
echo "new line" >> file.txt # append to file
```
## Input redirection (<)
```bash
sort < unsorted.txt         # take input from file
```
## Error redirection
```bash
command 2> error.log        # redirect errors only
command > output.log 2>&1   # redirect both output and errors
```
## Pipes (|) – THE MOST IMPORTANT CONCEPT
```bash
ls -la | grep ".txt"        # list and filter txt files
ps aux | grep python         # find python processes
cat file.txt | wc -l         # count lines (useless use of cat, but works)
history | grep git           # find all git commands you used
```

# 7. Permissions
## Understanding permissions
```bash
ls -l
-rwxr-xr-x  1 user group  1234 Jan 1 12:00 script.sh
```
```TEXT
││││││││││
││││││││└── others (read, execute)
│││││││└─── group (read, execute)
││││││└──── owner (read, write, execute)
│││││└───── other permissions
││││└────── group name
│││└─────── owner name
││└──────── number of links
│└───────── file size
└────────── file type (-=file, d=directory, l=link)
```
## Changing permissions
```bash
chmod +x script.sh          # add execute permission for all
chmod 755 script.sh         # rwxr-xr-x (owner all, group/others read+execute)
chmod 644 file.txt          # rw-r--r-- (owner read+write, others read)
```

# 8. Processes
## Viewing processes
```bash
ps                          # your processes
ps aux                      # all processes
top                         # interactive process viewer
htop                        # nicer top (if installed)
```
## Killing processes
```bash
kill PID                    # kill process by ID
kill -9 PID                 # force kill
killall firefox             # kill all firefox processes
```
## Background and foreground
```bash
command &                   # run in background
Ctrl+Z                      # suspend foreground process
bg                          # continue suspended process in background
fg                          # bring background process to foreground
jobs                        # list background jobs
```

# 9. Bash Basics – Scripting
## First script
**Create file** ***hello.sh:***

```bash
#!/bin/bash
echo "Hello, $USER!"
echo "Today is $(date)"
```
**Make it executable and run:**

```bash
chmod +x hello.sh
./hello.sh
```
**Variables**
```bash
name="John"
echo "My name is $name"
echo "My name is ${name}"   # same, but allows concatenation
```
**User input**
```bash
read -p "Enter your name: " name
echo "Hello, $name"
```
**Conditions**
```bash
if [ "$name" == "John" ]; then
    echo "Welcome, John!"
else
    echo "You are not John!"
fi
```
**Loops**
```bash
# for loop
for i in 1 2 3 4 5; do
    echo "Number: $i"
done

# while loop
counter=1
while [ $counter -le 5 ]; do
    echo "Counter: $counter"
    ((counter++))
done
```
**Functions**
```bash
greet() {
    echo "Hello, $1!"
}

greet "World"
```

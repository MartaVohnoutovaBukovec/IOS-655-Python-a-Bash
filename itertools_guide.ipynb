# The Complete Itertools Masterclass
This notebook contains practical, bug-free examples for every element in the `itertools` library.

## 1. Infinite Iterators
These generate streams that never end. We use `islice` or `break` to manage them.

```python
from itertools import count, cycle, repeat, islice

print("--- count(start, step) ---")
# Practical: Generating unique, incrementing IDs
counter = count(100, 10)
print("IDs:", [next(counter) for _ in range(5)])

print("\n--- cycle(iterable) ---")
# Practical: Creating an infinite loop of tasks or players
spinner = cycle(['-', '/', '|', '\\'])
print("Spinner:", [next(spinner) for _ in range(8)])

print("\n--- repeat(object, times) ---")
# Practical: Providing a constant stream of input to a function
print("Constant:", list(repeat('Data_Chunk', 3)))
```

## 2. Iterators Terminating on Shortest Input
Tools for processing data sequences efficiently.

```python
from itertools import (
    accumulate, chain, compress, dropwhile, 
    takewhile, filterfalse, groupby, islice, 
    starmap, tee, zip_longest
)
import operator

data = [1, 2, 3, 4, 1, 2]

print("1. accumulate:", list(accumulate(data, operator.add)))

print("2. chain:", list(chain(['A', 'B'], [1, 2])))

print("3. compress:", list(compress('ABCDE', [1, 0, 1, 0, 1])))

print("4. dropwhile (skip until >=3):", list(dropwhile(lambda x: x < 3, data)))

print("5. takewhile (stop at 4):", list(takewhile(lambda x: x < 4, data)))

print("6. filterfalse (keep odds):", list(filterfalse(lambda x: x % 2 == 0, range(10))))

print("7. groupby (requires sorted data):")
sorted_data = sorted([('fruit', 'apple'), ('fruit', 'banana'), ('veg', 'carrot')])
for key, group in groupby(sorted_data, lambda x: x[0]):
    print(f"   {key}: {list(group)}")

print("8. islice (indices 2 to 5):", list(islice(range(10), 2, 5)))

print("9. starmap (exponentiation):", list(starmap(pow, [(2, 5), (3, 2)])))

print("10. tee (cloning iterators):")
a, b = tee([1, 2, 3], 2)
print(f"   Stream A: {list(a)}, Stream B: {list(b)}")

print("11. zip_longest:", list(zip_longest([1, 2], ['a'], fillvalue='?')))
```

## 3. Combinatoric Iterators
Used for math, probability, and brute-force searching.

```python
from itertools import product, permutations, combinations, combinations_with_replacement

print("1. product (Cartesian):", list(product('AB', [1, 2])))

print("2. permutations (Order matters):", list(permutations('ABC', 2)))

print("3. combinations (Order doesn't matter):", list(combinations('ABC', 2)))

print("4. combinations_with_replacement:", list(combinations_with_replacement('ABC', 2)))
```
